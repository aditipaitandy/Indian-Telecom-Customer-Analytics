## Remove Negative Values

```python
df = df[
    (df["calls_made"] >= 0) &
    (df["sms_sent"] >= 0) &
    (df["data_used"] >= 0)
]
```

## Convert Date Column

```python
df["date_of_registration"] = pd.to_datetime(df["date_of_registration"])
```
